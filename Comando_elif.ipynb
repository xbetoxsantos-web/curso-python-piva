```python
Nota = float(input('Nota: '))
if Nota<0 or Nota>1:
  print('Entrada errada')
elif Nota>=0.9:
  print('Seu conceito é A')
elif Nota>=0.8:
  print('Seu conceito é B')
elif Nota>=0.7:
  print('Seu conceito é C')
elif Nota>=0.6:
  print('Seu conceito é D')
else:
  print('Seu conceito é E')
```

Faça um programa em python que...
Receba um valor inteiro, positivo, maior que zero.

Guarde esse valor em uma variável chamada: numero

Receba uma frase, e guarde essa frase na variável: texto.

Converta o número recebido para Decimal (float) e guarde esse novo valor convertido na variável x.

Coloque todos os caracteres da frase recebida em maiúscula, e guarde na mesma variável: texto.

Imprima as duas variáveis: texto e x.

```python
numero=input('Receba valor inteiro positivo: ')
texto=input('Receba uma frase: ')
x=float(numero)
texto=texto.upper()
print(texto)
print(x)
#
```

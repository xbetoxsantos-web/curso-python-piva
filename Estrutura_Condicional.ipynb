```python
idade = int(input('Qual a sua idade? '))
if idade >= 18:
  print('Parabéns você já é maior e idade')

else:
  print('Espere mais um pouco...')

```

```python
Nota1 = float(input('Digite a primeira nota: '))
Nota2 = float(input('Digite a segunda nota: '))
Média = (Nota1 + Nota2) / 2

if Média == 7:
    print('Parabéns você foi aprovado: ', Média)
elif Média >= 5 and Média <= 6.9:
    print('Dependência: ', Média)  # Adicionado o print com a recuo correto
elif Média > 7 and Média <= 10:
    print('aprovado: ', Média + 0.5)  # Adicionada a vírgula antes de Média
else:
    print('Reprovado: ', Média)
```
